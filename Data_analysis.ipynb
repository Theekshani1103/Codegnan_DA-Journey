```python
#updating the series
d = [1,2,3,4,5]
s = pd.Series(d)
print(s)
s[2]=90
s[6]=45
print(s)
s = s.drop(1)
print(s)
```

```python
#Inserting range of values
data = [9,8,7,6,5]
s = pd.Series(data,index=[0,1,2,3,4])
new_data = pd.Series([100,200,300],index=[5,6,7])
s = pd.concat([s,new_data])
print(s)
```

```python
#drop range:
import pandas as pd
data = [9,8,7,6,5]
s = pd.Series(data,index=[0,1,2,3,4])
s = s.drop(s.index[0:3])
print(s)
del s
```
